## 6. Historical 8-Engine Meta-Suite Assembly
The 8-Engine Meta-Suite combines diverse historical model architectures:
* **V17 Suite Members** (`v_17_full_suite_bundle.npz`): `oof_e1Aet`, `oof_e1et`, `final_oof`
* **V55 Suite Members** (`v55_members.npz`): `oof_cb_product`, `oof_deep_et`, `oof_v2_allfam_oh`, `oof_v2_mono`
* **Hist-XGBoost 10-Fold** (`oof_10f_xgb.npy`, `test_10f_xgb.npy`)

All predictions are mapped to standard normal quantiles via rank-probit transformation:
$$Z_k = \Phi^{-1}\left(\frac{\text{rank}(U_k) - 0.5}{N}\right)$$
and weighted by their verified Nelder-Mead simplex weights.


```python
t0_suite = time.time()
print("[*] Assembling Historical 8-Engine Meta-Suite...")

HIST_CACHE = Path("D:/gci/gemini/dependent/exp85_notebook_run/historical_cache")
if not HIST_CACHE.exists():
    HIST_CACHE = CACHE_DIR

v17_path   = HIST_CACHE / "v_17_full_suite_bundle.npz"
v55_path   = HIST_CACHE / "v55_members.npz"
xgb10_oof  = HIST_CACHE / "oof_10f_xgb.npy"
xgb10_test = HIST_CACHE / "test_10f_xgb.npy"

v17 = np.load(v17_path)
v55 = np.load(v55_path)
xgb_oof_arr  = np.load(xgb10_oof)
xgb_test_arr = np.load(xgb10_test)

def rk(x):
    return (rankdata(x, method="average") - 0.5) / len(x)

def to_probit(u, eps=1e-6):
    return norm.ppf(np.clip(u, eps, 1.0 - eps))

weights = {
    'v17_e1Aet':   0.2932,
    'xgb10':       0.2596,
    'v17_final':   0.2051,
    'v55_cbprod':  0.0874,
    'v17_e1et':    0.0489,
    'v55_deep_et': 0.0442,
    'v55_allfam':  0.0383,
    'v55_v2_mono': 0.0233
}
sum_w = sum(weights.values())
weights = {k: v / sum_w for k, v in weights.items()}

oof_models = {
    'v17_e1Aet':   rk(v17['oof_e1Aet']),
    'xgb10':       rk(xgb_oof_arr),
    'v17_final':   rk(v17['final_oof']),
    'v55_cbprod':  rk(v55['oof_cb_product']),
    'v17_e1et':    rk(v17['oof_e1et']),
    'v55_deep_et': rk(v55['oof_deep_et']),
    'v55_allfam':  rk(v55['oof_v2_allfam_oh']),
    'v55_v2_mono': rk(v55['oof_v2_mono'])
}

test_models = {
    'v17_e1Aet':   rk(v17['test_e1Aet']),
    'xgb10':       rk(xgb_test_arr),
    'v17_final':   rk(v17['final_test']),
    'v55_cbprod':  rk(v55['test_cb_product']),
    'v17_e1et':    rk(v17['test_e1et']),
    'v55_deep_et': rk(v55['test_deep_et']),
    'v55_allfam':  rk(v55['test_v2_allfam_oh']),
    'v55_v2_mono': rk(v55['test_v2_mono'])
}

oof_suite_z = np.zeros(n_train, dtype=np.float64)
test_suite_z = np.zeros(n_test, dtype=np.float64)

for k, w in weights.items():
    oof_suite_z  += w * to_probit(oof_models[k])
    test_suite_z += w * to_probit(test_models[k])

suite_oof_auc = roc_auc_score(y_train_raw, oof_suite_z)
print(f"[*] 8-Engine Suite OOF AUC: {suite_oof_auc:.6f}")
```

```python
t0_syn = time.time()
print("[*] Executing Probit Macro-Synthesis (85/15 Law)...")

OUT_DIR = Path("D:/gci/gemini/champion/cacherecreation")
OUT_DIR.mkdir(parents=True, exist_ok=True)
out_path = OUT_DIR / "submission_exp85_recovery_v5.csv"

# Track 2: Exact Historical Dependency Synthesis
z74_hist = to_probit(y_anchor_hist)
z_macro_hist = 0.850 * z74_hist + 0.150 * test_suite_z

ordinal_ranks_hist = rankdata(z_macro_hist, method="ordinal").astype(np.float64)
final_submission_target = (ordinal_ranks_hist - 0.5) / n_test

sub_df = pd.DataFrame({
    "SK_ID_CURR": test_ids_raw,
    "TARGET": final_submission_target
})

assert len(sub_df) == 61500, f"Expected 61,500 rows, got {len(sub_df)}"
assert sub_df["TARGET"].nunique() == 61500, "Ties detected in ranking!"
assert not sub_df.isna().any().any(), "Null values detected!"

sub_df.to_csv(out_path, index=False, float_format="%.10f")
print(f"[*] Exported Primary Submission : {out_path}")
print(f"    Size: {os.path.getsize(out_path) / 1e6:.2f} MB")

# Track 1: Rebuilt Upstream Synthesis (if available)
rebuilt_out_path = OUT_DIR / "submission_exp85_recovery_v5_rebuilt.csv"
if rebuilt_anchor_path.exists():
    z74_reb = to_probit(y_anchor_reb)
    z_macro_reb = 0.850 * z74_reb + 0.150 * test_suite_z
    ordinal_ranks_reb = rankdata(z_macro_reb, method="ordinal").astype(np.float64)
    final_reb_target = (ordinal_ranks_reb - 0.5) / n_test
    sub_reb_df = pd.DataFrame({"SK_ID_CURR": test_ids_raw, "TARGET": final_reb_target})
    sub_reb_df.to_csv(rebuilt_out_path, index=False, float_format="%.10f")
    print(f"[*] Exported Rebuilt Submission : {rebuilt_out_path}")
```
